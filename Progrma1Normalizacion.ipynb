**Normalización:** Proceso en el cual convertimos nuestros datos en un rango de 0 y 1 A este proceso se lo denimina homotecia.


**Formula:**

$x_i^N = \left\{ \frac{x_{ij} - \min(x_i)}{\max(x_i) - \min(x_i)}, \quad \forall j \in \{1,\ldots,n\} \right\}$


```python
from sklearn import preprocessing
import numpy as np
```

**sklearn.preprocessing:** permite preparar y transformar los datos antes de utilizarlos en modelos de aprendizaje automático.

**numpy:** permite trabajar con arreglos, matrices y operaciones matemáticas.

**as np:** es una abreviatura para utilizar NumPy de manera más sencilla.

```python
# Carga de datos.
X_train = np.array([[ 1., -4., 5.],
[ 5., 0., 2.],
[ 0., 6., -3.]])
print(X_train)
```

La carga de datos consiste en ingresar la información que será utilizada durante el procesamiento.

**¿Qué hace? Crea una matriz de tres filas y tres columnas con diferentes valores numéricos.**

**X_train:** almacena los datos de entrenamiento.

**np.array():** convierte los valores ingresados en una matriz de NumPy.

Cada fila representa un registro y cada columna una característica.

```python
# Limpieza de datos: normalización.
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

La normalización es un proceso que permite transformar los datos numéricos a un rango entre 0 y 1, manteniendo la proporción entre sus valores.

**¿Qué hace cada cosa?**

**- MinMaxScaler():** crea una herramienta que permite normalizar los datos.

**- normalizer:** variable que almacena la herramienta de normalización.
  
**- fit_transform():** identifica los valores mínimos y máximos de cada columna y realiza la transformación.
  
**- X_train_norm:** guarda los datos normalizados.

```python
# Salida de resultados.
print(X_train_norm)
```

**print():** muestra los resultados en pantalla.


**X_train_norm:** contiene la matriz con los valores normalizados.

```python
 # Ejemplo para ver que los resultados son iguales si aplicamos los métodos␣ "fit" y "transform" por separado
X_train_2 = np.copy(X_train)
# Crea un objeto "fitted_normalizer" llamando al método "fit"
f = normalizer.fit(X_train)
# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2"
X_train_norm_2 = f.transform(X_train)
# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

Los métodos fit() y transform() permiten realizar la normalización de datos en dos pasos, obteniendo el mismo resultado que con fit_transform().


**¿Qué hace cada cosa?**

**- np.copy():** crea una copia de los datos originales.

**- fit():** identifica los valores mínimos y máximos de cada columna.

**- transform():** transforma los datos utilizando los valores calculados anteriormente.

**- X_train_norm_2:** almacena la segunda matriz normalizada.

**- np.array_equiv():** compara las dos matrices para verificar si son iguales.
