**Estandarización:** La estandarización consiste en transformar la distribución de los valores de los atributos para que estos posean
una media y desviación típica determinadas. Esta transformación se lleva a cabo para cada atributo de forma
independiente.


**Formula:**

$x_j^s = \left\{ \frac{x_{ij} - \operatorname{media}(x_i)}{\operatorname{desv}(x_i)}, \ \forall j \in 1,\ldots,n \right\}$

```python
from sklearn import preprocessing
import numpy as np
```

**sklearn.preprocessing:** permite preparar y transformar los datos antes de utilizarlos en modelos de aprendizaje automático.

**numpy:** permite trabajar con arreglos, matrices y operaciones matemáticas.


**as np:** es una abreviatura para utilizar NumPy de manera más sencilla.

```python
# Carga de datos.
X_train = np.array([[ 1., -1., 2.],
[ 2., 0., 0.],
[ 0., 1., -1.]])
```

La carga de datos consiste en ingresar la información que será utilizada durante el procesamiento.


**¿Qué hace? Crea una matriz de tres filas y tres columnas con diferentes valores numéricos.**


**X_train:** almacena los datos de entrenamiento.


**np.array():** convierte los valores ingresados en una matriz de NumPy.


Cada fila representa un registro y cada columna una característica.

```python
# Limpieza de datos: estandarización.
standardizer = preprocessing.StandardScaler()
# Estandarizamos los datos de entrenamiento
X_train_std=standardizer.fit_transform(X_train)
print(X_train_std)
```

**StandardScaler():** crea una herramienta para estandarizar los datos.


**fit_transform():** calcula la media y la desviación estándar de cada columna y transforma sus valores.


**X_train_std:** guarda los datos estandarizados.


**print():** muestra los resultados en pantalla.
