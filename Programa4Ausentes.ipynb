```python
import numpy as np
from sklearn.impute import SimpleImputer
```


**Explicación:** Se importa NumPy para trabajar con datos numéricos y SimpleImputer para reemplazar los valores ausentes.

**Resultado:** Las librerías quedan disponibles para realizar la limpieza de datos.


```python
# Carga de datos
X_train = [[1, 4],
[np.nan, 3],
[7, 8]]
```


**Explicación:** Se crea una matriz de entrenamiento llamada X_train, que contiene valores numéricos y un dato ausente representado por np.nan.

**Resultado:** Se obtiene una matriz de 3 filas y 2 columnas, donde falta un valor en la primera columna.


```python
# Limpieza de datos: imputación valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizamos la estrategia de la media
# Aplicar los métodos "fit" y "transform" para imputar los valores pérdidos de␣X_train
X_train_clean = imputer.fit_transform(X_train)
print(X_train_clean)
```


**Explicación:** SimpleImputer(strategy='mean') calcula la media de cada columna para reemplazar los valores ausentes. fit_transform() aprende las medias y realiza el reemplazo.

**Resultado:** El valor ausente se reemplazó por 4, que es el promedio de 1 y 7. La matriz queda completa, sin datos faltantes.


```python
# Imputación de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
[6, np.nan],
[8, 2]]
1
# Imputar los datos del conjunto de test
X_test_clean=imputer.transform(X_test)
print(X_test_clean)
```


**Explicación:** Se utiliza imputer.transform() para reemplazar los valores ausentes del conjunto de prueba con las medias calculadas anteriormente en el entrenamiento.

**Resultado:** Los valores ausentes se reemplazaron por 4 en la primera columna y 5 en la segunda, obteniendo una matriz completa.
