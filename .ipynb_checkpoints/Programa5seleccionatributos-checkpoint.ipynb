```python
import numpy as np
from sklearn.feature_selection import VarianceThreshold
```


**Explicación:** Se importa NumPy para trabajar con datos numéricos y VarianceThreshold para seleccionar atributos según su variación.

**Resultado:** Las librerías quedan disponibles para analizar y seleccionar las columnas más variables del dataset.


```python
# Carga de datos.
X = [[0, 0, 1], [0, 1, 0], [1, 0, 0],
[0, 1, 1], [0, 1, 0], [0, 1, 1]]
print(X)
```


**Explicación:** Se crea una matriz llamada X con valores binarios (0 y 1), que será utilizada para analizar la variabilidad de sus atributos.

**Resultado:** Se obtiene una matriz de 6 registros y 3 columnas, que representan los datos originales.


```python
# Limpieza de datos: selección de atributos.
# fijamos un umbral (th) para decidir qué porcentaje mínimo deben variar los␣valores de una columna para considerarse
# una columna relevante a la hora de mantenerla en nuestro estudio
th = 0.2
sel = VarianceThreshold(th)
print(sel.fit_transform(X))
```


**Explicación:** Se establece un umbral de varianza de 0.2. VarianceThreshold elimina las columnas cuya varianza no supera ese valor y conserva las más variables.

**Resultado:** La matriz original de 3 columnas se redujo a 2, eliminando la primera columna por su baja varianza y conservando las otras dos.
