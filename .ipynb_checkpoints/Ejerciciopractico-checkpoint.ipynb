```python
#Eliminamos columnas innecesarias en la limpieza de datos como vemos en el dataframe la columna id no aporta
#nada a nuestro modelo por lo tanto lo eliminamos
if 'Id' in df.columns:
    df=df.drop(columns=["Id"])
```

```python
#Mostramos los detalles de nuesto dataset
df.info()
#Vrificamos si tenemos datos nulos si lo tenemos lo sumamos
df.isnull().sum()
```

- `df.columns`: Obtiene los nombres de las columnas.
- `if 'Id' in df.columns`: Comprueba si existe la columna Id.
- `df.drop(columns=["Id"])`: Elimina la columna Id.
- `df.info()`: Muestra la cantidad de registros, columnas y tipos de datos.
- `df.isnull()`: Identifica los valores vacíos.
- `.sum()`: Cuenta los valores vacíos de cada columna.

**Resultado:** Tenemos 150 registros, 5 columnas y ningún valor nulo.

```python
#Separamos los datos de entrada con nuestra variable objetivo
X=df.drop(columns=['Species'])
y=df['Species']
```

- `X = df.drop(columns=['Species'])`: Guarda las medidas de las flores sin incluir su especie.
- `y = df['Species']`: Guarda los nombres de las especies.

**Resultado:** Separamos los datos de entrada (X) y la variable que queremos identificar (y).

```python
#Exploramos o generamos un analis exploratorio del conjutno de datos
X.describe()
```

- `X.describe()`: Muestra un resumen estadístico de los datos.
- `count`: Cantidad de registros.
- `mean`: Promedio de los valores.
- `std`: Desviación estándar o dispersión.
- `min`: Valor mínimo.
- `25%`: Primer cuartil.
- `50%`: Mediana.
- `75%`: Tercer cuartil.
- `max`: Valor máximo.

**Resultado:** Conocemos los valores y la distribución de las medidas de las flores.

```python
#dibujamso los datos originales
sns.boxplot(data=X)
#Ponemos un titulo a la imagen
plt.title("Datos originales")
plt.show()
```

- `sns.boxplot(data=X)`: Crea un diagrama de cajas con las medidas de las flores.
- `plt.title("Datos originales")`: Coloca un título al gráfico.
- `plt.show()`: Muestra el gráfico en pantalla.

**Resultado:** Observamos que las medidas tienen diferentes rangos y algunos valores atípicos.

```python
#Normalizamos el conjunto de datos
normalizer = preprocessing.MinMaxScaler()
X_norm=normalizer.fit_transform(X)
X_norm = pd.DataFrame(X_norm, columns=X.columns)
X_norm.head()
```

- `preprocessing.MinMaxScaler()`: Crea una herramienta para normalizar los datos entre 0 y 1.
- `normalizer.fit_transform(X)`: Calcula los valores mínimos y máximos y transforma los datos.
- `pd.DataFrame(X_norm, columns=X.columns)`: Convierte los resultados en una tabla con los nombres originales.
- `X_norm.head()`: Muestra las primeras 5 filas normalizadas.

**Resultado:** Todas las medidas quedan en un rango de 0 a 1.

```python
X_norm.describe()
```

- `X_norm.describe()`: Muestra las estadísticas de los datos normalizados.
- `min`: Comprueba el valor mínimo.
- `max`: Comprueba el valor máximo.
- `mean`: Muestra el promedio después de normalizar.
- `std`: Indica la dispersión de los nuevos valores.

**Resultado:** Los valores mínimos son 0 y los máximos son 1, confirmando la normalización.

```python
#Estaddarización
standarizer = preprocessing.StandardScaler()
#Estandarizamos los datos
X_std=standarizer.fit_transform(X)
X_std= pd.DataFrame(X_std, columns=X.columns)
X_std.head()
```

- `preprocessing.StandardScaler()`: Crea una herramienta para estandarizar los datos.
- `standarizer.fit_transform(X)`: Calcula el promedio y la desviación estándar para transformar los valores.
- `pd.DataFrame(X_std, columns=X.columns)`: Organiza los valores estandarizados en una tabla.
- `X_std.head()`: Muestra las primeras 5 filas estandarizadas.

**Resultado:** Los datos quedan centrados alrededor de 0 y pueden tener valores positivos o negativos.

```python
#Comporación visual
plt.figure(figsize=(12,4))
```

```python
plt.subplot(1,3,1)
sns.boxplot(data=X)
plt.title("Original")
plt.subplot(1,3,2)
sns.boxplot(data=X_norm)
plt.title("Normalizado")
plt.subplot(1,3,3)
sns.boxplot(data=X_std)
plt.title("Estandarizado")
plt.show()
```

- `plt.figure(figsize=(12,4))`: Crea una figura de 12 pulgadas de ancho y 4 de alto.
- `plt.subplot(1,3,1)`: Selecciona la primera posición del gráfico.
- `sns.boxplot(data=X)`: Muestra los datos originales.
- `plt.title("Original")`: Coloca el título Original.
- `plt.subplot(1,3,2)`: Selecciona la segunda posición.
- `sns.boxplot(data=X_norm)`: Muestra los datos normalizados.
- `plt.title("Normalizado")`: Coloca el título Normalizado.
- `plt.subplot(1,3,3)`: Selecciona la tercera posición.
- `sns.boxplot(data=X_std)`: Muestra los datos estandarizados.
- `plt.title("Estandarizado")`: Coloca el título Estandarizado.
- `plt.show()`: Presenta los tres gráficos.

**Resultado:** Comparamos cómo cambian las escalas de los datos al aplicar normalización y estandarización.
